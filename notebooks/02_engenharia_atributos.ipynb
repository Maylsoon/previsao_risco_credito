## **Bibliotecas**

```python
import pandas as pd
import numpy as np
import joblib
```

```python
# carregando as bases
X_train, X_test, y_train, y_test = joblib.load('../bases/bases_modelagem/dados_modelagem.pkl')
```

## **Engenharia de atributos**

```python
# criando casados com casa própria
X_train['casado_casa_propria'] = ((X_train['estado_civil'] == 1) & (X_train['propriedade'] == 1)).astype(int)
X_test['casado_casa_propria'] = ((X_test['estado_civil'] == 1) & (X_test['propriedade'] == 1)).astype(int)

# criando faixas de renda: 0 - renda baixa, 1 - renda media, 2 - renda alta
q1 = X_train['renda'].quantile(0.25)
q2 = X_train['renda'].quantile(0.75)

X_train['faixa_renda'] = np.where(X_train['renda'] < q1, 0, np.where(X_train['renda'] < q2, 1, 2))
X_test['faixa_renda'] = np.where(X_test['renda'] < q1, 0, np.where(X_test['renda'] < q2, 1, 2))
```

```python
# apagando coluna ´renda´ que já foi transformada
X_train = X_train.drop(columns=['renda'])
X_test = X_test.drop(columns=['renda'])

# igualando estruturas 
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# Validação
print(f'base de treino final: {X_train.shape[0]} linhas | {X_train.shape[1]} colunas')
print(f'base de teste final: {X_test.shape[0]} linhas | {X_test.shape[1]} colunas')
```

```python
# Salvando para modelagem

joblib.dump((X_train, X_test, y_train, y_test), '../bases/bases_modelagem/dados_modelagem.pkl')
print("\n💾 Salvo em: bases/bases_modelagem/dados_modelagem.pkl2")
```
